```python
import numpy as np
import plotly.express as px
import pandas as pd
import plotly.io as pio
pio.renderers.default = "vscode"



medical_df = pd.read_csv('medical.csv')
smoker_values = {'no': 0, 'yes': 1}
smoker_numeric = medical_df.smoker.map(smoker_values)
medical_df.charges.corr(smoker_numeric)


non_smoker_df = medical_df[medical_df.smoker == 'no']


x_mean = non_smoker_df.age.mean()
y_mean = non_smoker_df.charges.mean()

coefficient = np.sum((non_smoker_df.age - x_mean) * (non_smoker_df.charges - y_mean)) / np.sum((non_smoker_df.age - x_mean) ** 2)
c = y_mean - coefficient * x_mean

def predict_charges(age, m=coefficient, b=c):
    return m * age + b

print(predict_charges(30))

x_test = np.linspace(18, 65, 47)
y_predicted = predict_charges(x_test)

fig = px.scatter(non_smoker_df, x='age', y='charges', color='sex', title='Medical Charges vs Age for Non-Smokers')
line_fig = px.line(x=x_test, y=y_predicted, labels={'x': 'Age', 'y': 'Charges'})
fig.add_traces(line_fig.data)
fig.show()
```
